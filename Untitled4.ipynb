```python
# =========================================================================
# CELL 1: Backbones (Split-ResNet and Tiny-ViT for Diamond Graph)
# =========================================================================
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
import copy

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# -------------------------------------------------------------------------
# 1. Split-ResNet (1-channel MNIST/Fashion-MNIST)
# -------------------------------------------------------------------------
class BasicBlock(nn.Module):
    def __init__(self, in_planes, planes, stride=1):
        super().__init__()
        self.conv1 = nn.Conv2d(in_planes, planes, kernel_size=3, stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(planes)
        self.conv2 = nn.Conv2d(planes, planes, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(planes)
        self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != planes:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_planes, planes, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(planes)
            )

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += self.shortcut(x)
        return F.relu(out)

class ResNetSource(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn = nn.BatchNorm2d(32)
        self.layer1 = BasicBlock(32, 32)
    def forward(self, x):
        # Outputs [B, 32, 28, 28] -> Split 16 channels to Worker 1, 16 to Worker 2
        return self.layer1(F.relu(self.bn(self.conv(x))))

class ResNetWorker(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer2 = BasicBlock(16, 32, stride=2) # [B, 32, 14, 14]
        self.pool = nn.AdaptiveAvgPool2d((1, 1))   # [B, 32]
    def forward(self, x):
        return torch.flatten(self.pool(self.layer2(x)), 1)

class ResNetSink(nn.Module):
    def __init__(self, in_dim=64, num_classes=10):
        super().__init__()
        self.classifier = nn.Linear(in_dim, num_classes)
    def forward(self, x):
        return self.classifier(x)

# -------------------------------------------------------------------------
# 2. Tiny-ViT (Patch size 4x4, Embedding 64)
# -------------------------------------------------------------------------
class ViTSource(nn.Module):
    def __init__(self, patch_size=4, emb_dim=64):
        super().__init__()
        self.proj = nn.Conv2d(1, emb_dim, kernel_size=patch_size, stride=patch_size)
        self.pos_embed = nn.Parameter(torch.zeros(1, 49, emb_dim))
        layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(layer, num_layers=1)
    def forward(self, x):
        x = self.proj(x).flatten(2).transpose(1, 2) + self.pos_embed
        return self.block(x) # [B, 49, 64]

class ViTWorker(nn.Module):
    def __init__(self, emb_dim=64):
        super().__init__()
        layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(layer, num_layers=1)
    def forward(self, x):
        # Return pooled token representation [B, 64]
        return self.block(x).mean(dim=1)

class ViTSink(nn.Module):
    def __init__(self, in_dim=128, num_classes=10):
        super().__init__()
        self.classifier = nn.Linear(in_dim, num_classes)
    def forward(self, x):
        return self.classifier(x)

print("--> Backbone splits defined.")
```

```python
# =========================================================================
# CELL 2: Codecs and Mutual Information Rate Bounds
# =========================================================================

class VariationalEncoder(nn.Module):
    def __init__(self, in_dim, bottleneck_dim):
        super().__init__()
        self.fc_mu = nn.Linear(in_dim, bottleneck_dim)
        self.fc_logvar = nn.Linear(in_dim, bottleneck_dim)

    def forward(self, z):
        mu = self.fc_mu(z)
        logvar = torch.clamp(self.fc_logvar(z), min=-5.0, max=5.0)
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        u = mu + eps * std
        return u, mu, logvar, eps

class PhysicalChannel(nn.Module):
    def __init__(self, snr_db=10.0):
        super().__init__()
        self.snr_db = snr_db
    def forward(self, u):
        norm = torch.norm(u, p=2, dim=-1, keepdim=True) + 1e-8
        u_norm = u / norm
        snr_lin = 10.0 ** (self.snr_db / 10.0)
        noise = torch.randn_like(u_norm) * np.sqrt(1.0 / (2.0 * snr_lin))
        return (u_norm + noise) * norm

def compute_rate_vub(mu, logvar):
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=-1)
    return torch.mean(kl)

def compute_rate_club(u, mu, logvar, eps):
    std = torch.exp(0.5 * logvar)
    pos_term = eps.pow(2).sum(dim=-1)
    u_neg = torch.roll(u, shifts=1, dims=0)
    neg_term = (((u_neg - mu) / std).pow(2)).sum(dim=-1)
    return 0.5 * torch.mean(neg_term - pos_term)

print("--> Codecs and Loss modules initialized.")
```

```python
# =========================================================================
# CELL 3: Diamond Model supporting ResNet and ViT
# =========================================================================

class UnifiedDiamondModel(nn.Module):
    def __init__(self, backbone_type="resnet", bottleneck_dim=16, snr_db=10.0):
        super().__init__()
        self.backbone_type = backbone_type
        self.bottleneck_dim = bottleneck_dim

        if backbone_type == "resnet":
            self.source = ResNetSource()
            self.worker1 = ResNetWorker()
            self.worker2 = ResNetWorker()
            worker_out_dim = 32
        elif backbone_type == "vit":
            self.source = ViTSource()
            self.worker1 = ViTWorker()
            self.worker2 = ViTWorker()
            worker_out_dim = 64
        else:
            raise ValueError(backbone_type)

        self.enc_w1 = VariationalEncoder(worker_out_dim, bottleneck_dim)
        self.enc_w2 = VariationalEncoder(worker_out_dim, bottleneck_dim)
        self.channel = PhysicalChannel(snr_db=snr_db)
        self.sink = nn.Linear(2 * bottleneck_dim, 10)

    def forward(self, x):
        if self.backbone_type == "resnet":
            z_s = self.source(x)
            z_w1 = self.worker1(z_s[:, :16])
            z_w2 = self.worker2(z_s[:, 16:])
        else:
            z_s = self.source(x)
            z_w1 = self.worker1(z_s[:, :25])
            z_w2 = self.worker2(z_s[:, 25:])

        u1, mu1, logvar1, eps1 = self.enc_w1(z_w1)
        u2, mu2, logvar2, eps2 = self.enc_w2(z_w2)

        u_bar1 = self.channel(u1)
        u_bar2 = self.channel(u2)

        r_t = torch.cat([u_bar1, u_bar2], dim=-1)
        y_hat = self.sink(r_t)

        edge_data = {
            "w1": {"u": u1, "mu": mu1, "logvar": logvar1, "eps": eps1},
            "w2": {"u": u2, "mu": mu2, "logvar": logvar2, "eps": eps2}
        }
        return y_hat, edge_data

print("--> Diamond Pipeline operational.")
```

```python
# =========================================================================
# CELL 4: Training Engine (Joint E2E, Two-Stage Frozen, and Fine-Tuning)
# =========================================================================

# Helper to pretrain backbones once (saves huge amounts of runtime)
def pretrain_backbone(backbone_type="resnet", epochs=5):
    print(f"--> Pretraining baseline backbone: {backbone_type.upper()}...")
    model = UnifiedDiamondModel(backbone_type=backbone_type, bottleneck_dim=32, snr_db=30.0).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

    for epoch in range(epochs):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            y_hat, _ = model(x)
            loss = F.cross_entropy(y_hat, y)
            loss.backward()
            optimizer.step()

    # Extract only backbone weights
    backbone_weights = {
        "source": copy.deepcopy(model.source.state_dict()),
        "worker1": copy.deepcopy(model.worker1.state_dict()),
        "worker2": copy.deepcopy(model.worker2.state_dict())
    }
    print(f"--> {backbone_type.upper()} Pretraining complete.")
    return backbone_weights

# Execution Engine
def train_and_eval(model, pretrained_weights, training_style, rate_type, epochs=5, lam=1e-3):
    model = model.to(device)

    # 1. Apply training style logic
    if training_style == "jointly":
        # Train everything end-to-end from scratch
        optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

    elif training_style == "separately":
        # Load pretrained backbone and freeze it completely
        model.source.load_state_dict(pretrained_weights["source"])
        model.worker1.load_state_dict(pretrained_weights["worker1"])
        model.worker2.load_state_dict(pretrained_weights["worker2"])
        for p in model.source.parameters(): p.requires_grad = False
        for p in model.worker1.parameters(): p.requires_grad = False
        for p in model.worker2.parameters(): p.requires_grad = False

        # Train only codecs + sink classifier
        trainable = list(model.enc_w1.parameters()) + list(model.enc_w2.parameters()) + list(model.sink.parameters())
        optimizer = torch.optim.Adam(trainable, lr=1e-3)

    elif training_style == "fine_tuned":
        # Load pretrained backbone and warm-start with differential learning rate
        model.source.load_state_dict(pretrained_weights["source"])
        model.worker1.load_state_dict(pretrained_weights["worker1"])
        model.worker2.load_state_dict(pretrained_weights["worker2"])
        backbone_p = list(model.source.parameters()) + list(model.worker1.parameters()) + list(model.worker2.parameters())
        codec_p = list(model.enc_w1.parameters()) + list(model.enc_w2.parameters()) + list(model.sink.parameters())
        optimizer = torch.optim.Adam([
            {"params": backbone_p, "lr": 1e-4},
            {"params": codec_p, "lr": 1e-3}
        ])

    # 2. Training Loop
    for epoch in range(epochs):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            y_hat, edge_data = model(x)

            # Loss: Task Cross Entropy + lambda * Rate Bound
            loss_task = F.cross_entropy(y_hat, y)
            loss_rate = 0.0
            for k in ["w1", "w2"]:
                if rate_type == "VUB":
                    loss_rate += compute_rate_vub(edge_data[k]["mu"], edge_data[k]["logvar"])
                else:
                    loss_rate += compute_rate_club(edge_data[k]["u"], edge_data[k]["mu"], edge_data[k]["logvar"], edge_data[k]["eps"])

            total_loss = loss_task + lam * loss_rate
            total_loss.backward()
            optimizer.step()

    # 3. Evaluation on Test Set
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            y_hat, _ = model(x)
            preds = torch.argmax(y_hat, dim=-1)
            correct += (preds == y).sum().item()
            total += x.size(0)

    return 100.0 * correct / total

print("--> Training and evaluation engine ready.")
```

```python
# =========================================================================
# CELL 5: Full Evaluation Grid (Strictly Aligned Rate Points)
# =========================================================================

# 1. Load Data
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081,))])
train_loader = DataLoader(datasets.MNIST(root="./data", train=True, download=True, transform=transform), batch_size=128, shuffle=True)
test_loader = DataLoader(datasets.MNIST(root="./data", train=False, download=True, transform=transform), batch_size=128, shuffle=False)

# 2. Pretrain each backbone once
pretrained_resnet = pretrain_backbone("resnet", epochs=4)
pretrained_vit = pretrain_backbone("vit", epochs=4)

# 3. Strictly Aligned Rate Points (Bottleneck Channel Dimensions)
rate_points = [4, 8, 16, 32, 64] # Aligned X-axis points

models = ["resnet", "vit"]
styles = ["jointly", "separately", "fine_tuned"]
pretrained_dict = {"resnet": pretrained_resnet, "vit": pretrained_vit}

# Create a 2x3 Grid Plot
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(18, 10), sharex=True)

for r_idx, model_name in enumerate(models):
    for c_idx, style_name in enumerate(styles):
        ax = axes[r_idx, c_idx]
        acc_vub = []
        acc_club = []

        print(f"\nEvaluating: Model={model_name.upper()} | Style={style_name.upper()}")
        for K in rate_points:
            # Evaluate VUB at Rate = K
            m_vub = UnifiedDiamondModel(backbone_type=model_name, bottleneck_dim=K, snr_db=10.0)
            a_v = train_and_eval(m_vub, pretrained_dict[model_name], style_name, rate_type="VUB", epochs=4)
            acc_vub.append(a_v)

            # Evaluate CLUB at Rate = K
            m_club = UnifiedDiamondModel(backbone_type=model_name, bottleneck_dim=K, snr_db=10.0)
            a_c = train_and_eval(m_club, pretrained_dict[model_name], style_name, rate_type="CLUB", epochs=4)
            acc_club.append(a_c)

            print(f"  [Rate K={K:2d}] --> VUB: {a_v:.2f}% | CLUB: {a_c:.2f}%")

        # Plot strictly aligned vertical points
        ax.plot(rate_points, acc_club, marker='o', linewidth=2, color='crimson', label="Proposed CLUB")
        ax.plot(rate_points, acc_vub, marker='s', linewidth=2, linestyle='--', color='navy', label="Baseline VUB")

        ax.set_title(f"{model_name.upper()} ({style_name.capitalize()})", fontsize=12, fontweight='bold')
        ax.set_xticks(rate_points)
        ax.grid(True, linestyle=":", alpha=0.6)
        if c_idx == 0:
            ax.set_ylabel("Target Accuracy (%)", fontsize=11)
        if r_idx == 1:
            ax.set_xlabel("Channel Rate (Bottleneck Dim $K$)", fontsize=11)
        ax.legend(fontsize=9, loc='lower right')

plt.suptitle("Rate-Accuracy Trade-off Across Backbones and Training Schemes (Diamond Topology)", fontsize=15, y=0.98)
plt.tight_layout()
plt.show()
```

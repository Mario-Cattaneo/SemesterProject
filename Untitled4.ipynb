```python
# =========================================================================
# CELL 1: Backbones (Split-ResNet and Tiny-ViT for Diamond Graph)
# =========================================================================
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
import copy

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# -------------------------------------------------------------------------
# 1. Split-ResNet (1-channel MNIST/Fashion-MNIST)
# -------------------------------------------------------------------------
class BasicBlock(nn.Module):
    def __init__(self, in_planes, planes, stride=1):
        super().__init__()
        self.conv1 = nn.Conv2d(in_planes, planes, kernel_size=3, stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(planes)
        self.conv2 = nn.Conv2d(planes, planes, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(planes)
        self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != planes:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_planes, planes, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(planes)
            )

    def forward(self, x):
        out = F.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += self.shortcut(x)
        return F.relu(out)

class ResNetSource(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn = nn.BatchNorm2d(32)
        self.layer1 = BasicBlock(32, 32)
    def forward(self, x):
        # Outputs [B, 32, 28, 28] -> Split 16 channels to Worker 1, 16 to Worker 2
        return self.layer1(F.relu(self.bn(self.conv(x))))

class ResNetWorker(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer2 = BasicBlock(16, 32, stride=2) # [B, 32, 14, 14]
        self.pool = nn.AdaptiveAvgPool2d((1, 1))   # [B, 32]
    def forward(self, x):
        return torch.flatten(self.pool(self.layer2(x)), 1)

class ResNetSink(nn.Module):
    def __init__(self, in_dim=64, num_classes=10):
        super().__init__()
        self.classifier = nn.Linear(in_dim, num_classes)
    def forward(self, x):
        return self.classifier(x)

# -------------------------------------------------------------------------
# 2. Tiny-ViT (Patch size 4x4, Embedding 64)
# -------------------------------------------------------------------------
class ViTSource(nn.Module):
    def __init__(self, patch_size=4, emb_dim=64):
        super().__init__()
        self.proj = nn.Conv2d(1, emb_dim, kernel_size=patch_size, stride=patch_size)
        self.pos_embed = nn.Parameter(torch.zeros(1, 49, emb_dim))
        layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(layer, num_layers=1)
    def forward(self, x):
        x = self.proj(x).flatten(2).transpose(1, 2) + self.pos_embed
        return self.block(x) # [B, 49, 64]

class ViTWorker(nn.Module):
    def __init__(self, emb_dim=64):
        super().__init__()
        layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(layer, num_layers=1)
    def forward(self, x):
        # Return pooled token representation [B, 64]
        return self.block(x).mean(dim=1)

class ViTSink(nn.Module):
    def __init__(self, in_dim=128, num_classes=10):
        super().__init__()
        self.classifier = nn.Linear(in_dim, num_classes)
    def forward(self, x):
        return self.classifier(x)

print("--> Backbone splits defined.")
```

```python
# =========================================================================
# CELL 2: Codecs and Mutual Information Rate Bounds
# =========================================================================

class VariationalEncoder(nn.Module):
    def __init__(self, in_dim, bottleneck_dim):
        super().__init__()
        self.fc_mu = nn.Linear(in_dim, bottleneck_dim)
        self.fc_logvar = nn.Linear(in_dim, bottleneck_dim)

    def forward(self, z):
        mu = self.fc_mu(z)
        logvar = torch.clamp(self.fc_logvar(z), min=-5.0, max=5.0)
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        u = mu + eps * std
        return u, mu, logvar, eps

class PhysicalChannel(nn.Module):
    def __init__(self, snr_db=10.0):
        super().__init__()
        self.snr_db = snr_db
    def forward(self, u):
        norm = torch.norm(u, p=2, dim=-1, keepdim=True) + 1e-8
        u_norm = u / norm
        snr_lin = 10.0 ** (self.snr_db / 10.0)
        noise = torch.randn_like(u_norm) * np.sqrt(1.0 / (2.0 * snr_lin))
        return (u_norm + noise) * norm

def compute_rate_vub(mu, logvar):
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=-1)
    return torch.mean(kl)

def compute_rate_club(u, mu, logvar, eps):
    std = torch.exp(0.5 * logvar)
    pos_term = eps.pow(2).sum(dim=-1)
    u_neg = torch.roll(u, shifts=1, dims=0)
    neg_term = (((u_neg - mu) / std).pow(2)).sum(dim=-1)
    return 0.5 * torch.mean(neg_term - pos_term)

print("--> Codecs and Loss modules initialized.")
```

```python
# =========================================================================
# CELL 3: Modular Diamond Architecture (Two-Stage Codecs & Channel)
# =========================================================================

class UnifiedDiamondModel(nn.Module):
    def __init__(self, backbone_type="resnet", bottleneck_dim=16, snr_db=10.0):
        super().__init__()
        self.backbone_type = backbone_type
        self.bottleneck_dim = bottleneck_dim

        if backbone_type == "resnet":
            self.source = ResNetSource()
            self.worker1 = ResNetWorker()
            self.worker2 = ResNetWorker()
            self.worker_dim = 32
        elif backbone_type == "vit":
            self.source = ViTSource()
            self.worker1 = ViTWorker()
            self.worker2 = ViTWorker()
            self.worker_dim = 64

        # Encoders and Channel
        self.enc_w1 = VariationalEncoder(self.worker_dim, bottleneck_dim)
        self.enc_w2 = VariationalEncoder(self.worker_dim, bottleneck_dim)
        self.channel = PhysicalChannel(snr_db=snr_db)

        # 1. Heads for JG: Joint classifier on [u1, u2]
        self.sink_jg = nn.Linear(2 * bottleneck_dim, 10)

        # 2. Heads for SG: Individual classifiers on u1 and u2
        self.head_sg1 = nn.Linear(bottleneck_dim, 10)
        self.head_sg2 = nn.Linear(bottleneck_dim, 10)

        # 3. Decoders for SL: Separate reconstruction of z_w1 and z_w2
        self.dec_sl1 = nn.Sequential(nn.Linear(bottleneck_dim, self.worker_dim), nn.ReLU(), nn.Linear(self.worker_dim, self.worker_dim))
        self.dec_sl2 = nn.Sequential(nn.Linear(bottleneck_dim, self.worker_dim), nn.ReLU(), nn.Linear(self.worker_dim, self.worker_dim))

        # 4. Decoder for JL: Joint reconstruction of [z_w1, z_w2]
        self.dec_jl = nn.Sequential(nn.Linear(2 * bottleneck_dim, 2 * self.worker_dim), nn.ReLU(), nn.Linear(2 * self.worker_dim, 2 * self.worker_dim))

        # Pretrained sink head for evaluating SL and JL reconstructions
        self.sink_pretrained = nn.Linear(2 * self.worker_dim, 10)

    def extract_workers(self, x):
        """Extract intermediate features z_w1, z_w2 from frozen backbone"""
        if self.backbone_type == "resnet":
            z_s = self.source(x)
            z_w1 = self.worker1(z_s[:, :16])
            z_w2 = self.worker2(z_s[:, 16:])
        else:
            z_s = self.source(x)
            z_w1 = self.worker1(z_s[:, :25])
            z_w2 = self.worker2(z_s[:, 25:])
        return z_w1, z_w2

    def forward_from_z(self, z_w1, z_w2):
        """Fast forward pass starting directly from intermediate worker representations"""
        u1, mu1, logvar1, eps1 = self.enc_w1(z_w1)
        u2, mu2, logvar2, eps2 = self.enc_w2(z_w2)

        u_bar1 = self.channel(u1)
        u_bar2 = self.channel(u2)
        r_t = torch.cat([u_bar1, u_bar2], dim=-1)

        # 1. JG Prediction
        y_hat_jg = self.sink_jg(r_t)

        # 2. SG Prediction
        y_hat_sg = 0.5 * (self.head_sg1(u_bar1) + self.head_sg2(u_bar2))

        # 3. SL Reconstructions
        z_recon_sl1 = self.dec_sl1(u_bar1)
        z_recon_sl2 = self.dec_sl2(u_bar2)
        y_hat_sl = self.sink_pretrained(torch.cat([z_recon_sl1, z_recon_sl2], dim=-1))

        # 4. JL Reconstruction
        z_recon_jl = self.dec_jl(r_t)
        y_hat_jl = self.sink_pretrained(z_recon_jl)

        edge_data = {
            "w1": {"z": z_w1, "u": u1, "mu": mu1, "logvar": logvar1, "eps": eps1, "z_recon_sl": z_recon_sl1},
            "w2": {"z": z_w2, "u": u2, "mu": mu2, "logvar": logvar2, "eps": eps2, "z_recon_sl": z_recon_sl2},
            "jl": {"z_true": torch.cat([z_w1, z_w2], dim=-1), "z_recon": z_recon_jl}
        }
        preds = {"JG": y_hat_jg, "SG": y_hat_sg, "JL": y_hat_jl, "SL": y_hat_sl}
        return preds, edge_data

print("--> Modular Diamond Architecture initialized.")
```

```python
# =========================================================================
# CELL 4: Ultra-Fast Two-Stage Training Engine (Cached Features)
# =========================================================================

def pretrain_and_cache_backbone(backbone_type="resnet", epochs=2):
    print(f"\n[Stage 1] Pretraining baseline {backbone_type.upper()} backbone (Epochs={epochs})...")
    model = UnifiedDiamondModel(backbone_type=backbone_type, bottleneck_dim=32, snr_db=30.0).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1.5e-3)
    scaler = torch.amp.GradScaler(device.type, enabled=(device.type == 'cuda'))

    for epoch in range(1, epochs + 1):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
            optimizer.zero_grad()
            with torch.amp.autocast(device_type=device.type, enabled=(device.type == 'cuda')):
                z_w1, z_w2 = model.extract_workers(x)
                y_hat = model.sink_pretrained(torch.cat([z_w1, z_w2], dim=-1))
                loss = F.cross_entropy(y_hat, y)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

    # Pre-extract intermediate features into RAM/VRAM
    print(f"  --> Pre-extracting and caching {backbone_type.upper()} worker tensors...")
    model.eval()
    cached_train, cached_test = [], []
    with torch.no_grad():
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            z1, z2 = model.extract_workers(x)
            cached_train.append((z1.cpu(), z2.cpu(), y.cpu()))
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            z1, z2 = model.extract_workers(x)
            cached_test.append((z1.cpu(), z2.cpu(), y.cpu()))

    # Evaluate Uncompressed Baseline (CR = 1x) over the physical channel
    print(f"  --> Measuring Uncompressed Baseline (CR = 1x) over physical channel...")
    channel = PhysicalChannel(snr_db=10.0).to(device)
    correct, total = 0, 0
    with torch.no_grad():
        for z1, z2, y in cached_test:
            z1, z2, y = z1.to(device), z2.to(device), y.to(device)
            z_bar1 = channel(z1)
            z_bar2 = channel(z2)
            y_hat = model.sink_pretrained(torch.cat([z_bar1, z_bar2], dim=-1))
            preds = torch.argmax(y_hat, dim=-1)
            correct += (preds == y).sum().item()
            total += y.size(0)
    uncompressed_acc = 100.0 * correct / total
    print(f"  --> {backbone_type.upper()} Uncompressed Baseline Accuracy: {uncompressed_acc:.2f}%")

    return model.state_dict(), cached_train, cached_test, uncompressed_acc

def train_codecs_on_cache(
    model, pretrained_state, cached_data,
    rel_type="JG", rate_type="CLUB", epochs=3, lam=1e-3
):
    model = model.to(device)
    scaler = torch.amp.GradScaler(device.type, enabled=(device.type == 'cuda'))

    # Load pretrained sink classifier for SL/JL evaluation
    sink_pretrained_weights = {
        k.replace("sink_pretrained.", ""): v
        for k, v in pretrained_state.items()
        if "sink_pretrained." in k
    }
    model.sink_pretrained.load_state_dict(sink_pretrained_weights)

    train_cache, test_cache = cached_data
    optimizer = torch.optim.Adam(model.parameters(), lr=2e-3)

    for epoch in range(epochs):
        model.train()
        for z1, z2, y in train_cache:
            z1, z2, y = z1.to(device), z2.to(device), y.to(device)
            optimizer.zero_grad()
            with torch.amp.autocast(device_type=device.type, enabled=(device.type == 'cuda')):
                preds, edge_data = model.forward_from_z(z1, z2)

                # Relevance Loss
                if rel_type == "JG":
                    loss_rel = F.cross_entropy(preds["JG"], y)
                elif rel_type == "SG":
                    loss_rel = 0.5 * (F.cross_entropy(model.head_sg1(edge_data["w1"]["u"]), y) +
                                      F.cross_entropy(model.head_sg2(edge_data["w2"]["u"]), y))
                elif rel_type == "SL":
                    loss_rel = F.mse_loss(edge_data["w1"]["z_recon_sl"], z1) + \
                               F.mse_loss(edge_data["w2"]["z_recon_sl"], z2)
                elif rel_type == "JL":
                    loss_rel = F.mse_loss(edge_data["jl"]["z_recon"], edge_data["jl"]["z_true"])

                # Rate Loss
                loss_rate = 0.0
                for k in ["w1", "w2"]:
                    if rate_type == "VUB":
                        loss_rate += compute_rate_vub(edge_data[k]["mu"], edge_data[k]["logvar"])
                    else:
                        loss_rate += compute_rate_club(edge_data[k]["u"], edge_data[k]["mu"], edge_data[k]["logvar"], edge_data[k]["eps"])

                total_loss = loss_rel + lam * loss_rate

            scaler.scale(total_loss).backward()
            scaler.step(optimizer)
            scaler.update()

    # Evaluation
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for z1, z2, y in test_cache:
            z1, z2, y = z1.to(device), z2.to(device), y.to(device)
            preds, _ = model.forward_from_z(z1, z2)
            correct += (torch.argmax(preds[rel_type], dim=-1) == y).sum().item()
            total += y.size(0)
    return 100.0 * correct / total

print("--> Fast Two-Stage Training Engine initialized.")
```

```python
# =========================================================================
# CELL 5: Compression Ratio Benchmark with Real-Time Progressive Drive Saving
# =========================================================================
import time

# 1. DataLoader
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081,))])
train_loader = DataLoader(datasets.MNIST(root="./data", train=True, download=True, transform=transform), batch_size=256, shuffle=True, pin_memory=(device.type == 'cuda'))
test_loader = DataLoader(datasets.MNIST(root="./data", train=False, download=True, transform=transform), batch_size=256, shuffle=False, pin_memory=(device.type == 'cuda'))

# 2. Pretrain & cache features once for ResNet and ViT
state_res, cache_tr_res, cache_te_res, base_acc_res = pretrain_and_cache_backbone("resnet", epochs=2)
state_vit, cache_tr_vit, cache_te_vit, base_acc_vit = pretrain_and_cache_backbone("vit", epochs=2)

model_configs = {
    "resnet": {"state": state_res, "cache": (cache_tr_res, cache_te_res), "base_acc": base_acc_res, "raw_dim": 32},
    "vit": {"state": state_vit, "cache": (cache_tr_vit, cache_te_vit), "base_acc": base_acc_vit, "raw_dim": 64}
}

# 3. Strictly Aligned Compression Ratios (CR)
# CR = 1x (Raw uncompressed baseline); CR > 1x (Compressed points)
compression_ratios = [1, 2, 4, 8, 16]
cr_ticks = [1, 2, 4, 8, 16]
cr_labels = ["1x\n(Raw)", "2x", "4x", "8x", "16x"]

# 4. Styling Definitions
color_map = {"CLUB": "crimson", "VUB": "navy"}
line_map = {"CLUB": "-", "VUB": "--"}
marker_map = {"JG": "o", "SG": "s", "JL": "^", "SL": "v"}

objectives = ["JG", "SG", "JL", "SL"]
bounds = ["CLUB", "VUB"]

total_runs = 2 * len(objectives) * len(bounds) * (len(compression_ratios) - 1)
run_idx = 0
start_time = time.time()
raw_metrics_log = {}

# Filepaths for saving in Google Drive
png_path = os.path.join(SAVE_DIR, "cr_sweep_two_stage_1x2.png")
pdf_path = os.path.join(SAVE_DIR, "cr_sweep_two_stage_1x2.pdf")
json_path = os.path.join(SAVE_DIR, "cr_sweep_metrics.json")

# 1x2 Subplot Grid (Left: ResNet, Right: Tiny-ViT)
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(16, 6), sharey=False)

print(f"\n{'='*80}")
print(f"--> STARTING TWO-STAGE COMPRESSION RATIO SWEEP (Total runs: {total_runs})")
print(f"{'='*80}")

for m_idx, (m_name, m_data) in enumerate(model_configs.items()):
    ax = axes[m_idx]
    raw_metrics_log[m_name] = {}
    base_acc = m_data["base_acc"]
    raw_dim = m_data["raw_dim"]

    for obj in objectives:
        for bnd in bounds:
            cr_curve = [base_acc]
            series_key = f"{obj}_{bnd}"

            # Sweep compressed ratios (2x, 4x, 8x, 16x)
            for CR in compression_ratios[1:]:
                run_idx += 1
                t0 = time.time()

                K = max(2, raw_dim // CR)

                model = UnifiedDiamondModel(backbone_type=m_name, bottleneck_dim=K, snr_db=10.0)
                acc = train_codecs_on_cache(
                    model, m_data["state"], m_data["cache"],
                    rel_type=obj, rate_type=bnd, epochs=3, lam=1e-3
                )
                cr_curve.append(acc)
                dt = time.time() - t0

                pct = (run_idx / total_runs) * 100.0
                print(f"[{run_idx:2d}/{total_runs} | {pct:5.1f}%] {m_name.upper():6s} | {obj:2s}-{bnd:4s} | "
                      f"CR={CR:2d}x (K={K:2d}) --> Acc: {acc:6.2f}% | Time: {dt:3.1f}s")

            raw_metrics_log[m_name][series_key] = cr_curve

            # Plot curve branching from CR=1x
            ax.plot(
                compression_ratios, cr_curve,
                color=color_map[bnd],
                marker=marker_map[obj],
                linestyle=line_map[bnd],
                linewidth=2.0,
                markersize=7,
                label=f"{obj}-{bnd}"
            )

    # Reference line for uncompressed baseline
    ax.axhline(y=base_acc, color="gray", linestyle=":", alpha=0.7, label=f"Uncompressed ({base_acc:.1f}%)")

    ax.set_title(f"{m_name.upper()} (Modular Two-Stage)", fontsize=13, fontweight="bold")
    ax.set_xticks(cr_ticks)
    ax.set_xticklabels(cr_labels, fontsize=10)
    ax.set_xlabel("Compression Ratio (Bandwidth Savings)", fontsize=11)
    ax.set_ylabel("Target Accuracy (%)", fontsize=11)
    ax.grid(True, linestyle=":", alpha=0.6)
    ax.legend(fontsize=9, loc="lower left")

    # PROGRESSIVE SAVE: Save immediately after ResNet finishes (so Drive is never empty)
    plt.savefig(png_path, dpi=300, bbox_inches='tight')
    plt.savefig(pdf_path, bbox_inches='tight')
    with open(json_path, 'w') as f:
        json.dump(raw_metrics_log, f, indent=4)
    print(f"  [Auto-Save] Updated results saved to Google Drive.")

plt.suptitle("Rate-Accuracy Trade-off vs. Compression Ratio (Two-Stage Frozen Backbone)", fontsize=14, y=1.02)
plt.tight_layout()

# Final save
plt.savefig(png_path, dpi=300, bbox_inches='tight')
plt.savefig(pdf_path, bbox_inches='tight')

total_mins = (time.time() - start_time) / 60.0
print(f"\n--> Benchmark successfully completed in {total_mins:.1f} minutes!")
print(f"    - PNG:  {png_path}")
print(f"    - PDF:  {pdf_path}")
print(f"    - JSON: {json_path}")

plt.show()
```

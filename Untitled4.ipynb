```python
# =========================================================================
# CELL 1: Imports, Device Setup, and Reproducibility
# =========================================================================
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets, transforms
import numpy as np
import matplotlib.pyplot as plt
import copy

# Set hardware device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"--> Using compute device: {device}")

# Set random seed for reproducibility
def set_seed(seed=42):
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    np.random.seed(seed)

set_seed(42)
```

```python
# =========================================================================
# CELL 2: Datasets & Topological Partitioning (Diamond & Tree)
# =========================================================================
def get_dataloaders(dataset_name="MNIST", batch_size=128):
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.1307,), (0.3081,))
    ])

    if dataset_name == "MNIST":
        train_data = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
        test_data = datasets.MNIST(root="./data", train=False, download=True, transform=transform)
    elif dataset_name == "FashionMNIST":
        train_data = datasets.FashionMNIST(root="./data", train=True, download=True, transform=transform)
        test_data = datasets.FashionMNIST(root="./data", train=False, download=True, transform=transform)
    else:
        raise ValueError(f"Unknown dataset: {dataset_name}")

    train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True, drop_last=True)
    test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False, drop_last=False)
    return train_loader, test_loader

# Helper function to partition inputs for Tree Topology (Split left/right 28x14)
def partition_tree_input(x):
    # x shape: [B, 1, 28, 28] -> x1: [B, 1, 28, 14], x2: [B, 1, 28, 14]
    x1 = x[:, :, :, :14]
    x2 = x[:, :, :, 14:]
    return x1, x2

print("--> Data pipeline ready.")
```

```python
# =========================================================================
# CELL 3: Modular Backbone Micro-Splits
# =========================================================================

# -------------------------------------------------------------------------
# 1. Multi-Stream MLP Splits
# -------------------------------------------------------------------------
class MLPSource(nn.Module):
    def __init__(self, in_dim=784, out_dim=256):
        super().__init__()
        self.fc = nn.Sequential(nn.Flatten(), nn.Linear(in_dim, out_dim), nn.ReLU())
    def forward(self, x): return self.fc(x)

class MLPWorker(nn.Module):
    def __init__(self, in_dim=128, out_dim=128):
        super().__init__()
        self.fc = nn.Sequential(nn.Linear(in_dim, out_dim), nn.ReLU())
    def forward(self, x): return self.fc(x)

class MLPSink(nn.Module):
    def __init__(self, in_dim=256, num_classes=10):
        super().__init__()
        self.classifier = nn.Linear(in_dim, num_classes)
    def forward(self, x): return self.classifier(x)


# -------------------------------------------------------------------------
# 2. Tiny-ViT Micro-Splits (Patch size 4x4, 4 Transformer Layers)
# -------------------------------------------------------------------------
class PatchEmbedding(nn.Module):
    def __init__(self, in_channels=1, patch_size=4, emb_dim=64, img_size=(28, 28)):
        super().__init__()
        self.patch_size = patch_size
        self.proj = nn.Conv2d(in_channels, emb_dim, kernel_size=patch_size, stride=patch_size)
        num_patches = (img_size[0] // patch_size) * (img_size[1] // patch_size)
        self.pos_embed = nn.Parameter(torch.zeros(1, num_patches, emb_dim))

    def forward(self, x):
        # x: [B, C, H, W] -> [B, emb_dim, H/P, W/P] -> [B, num_patches, emb_dim]
        B = x.shape[0]
        x = self.proj(x).flatten(2).transpose(1, 2)
        x = x + self.pos_embed[:, :x.size(1), :]
        return x

class ViTSource(nn.Module):
    def __init__(self, in_channels=1, emb_dim=64, img_size=(28, 28)):
        super().__init__()
        self.patch_embed = PatchEmbedding(in_channels=in_channels, emb_dim=emb_dim, img_size=img_size)
        encoder_layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(encoder_layer, num_layers=1)
    def forward(self, x):
        return self.block(self.patch_embed(x))

class ViTWorker(nn.Module):
    def __init__(self, emb_dim=64):
        super().__init__()
        encoder_layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(encoder_layer, num_layers=1)
    def forward(self, x):
        return self.block(x)

class ViTSink(nn.Module):
    def __init__(self, emb_dim=128, num_classes=10):
        super().__init__()
        encoder_layer = nn.TransformerEncoderLayer(d_model=emb_dim, nhead=4, dim_feedforward=128, batch_first=True)
        self.block = nn.TransformerEncoder(encoder_layer, num_layers=1)
        self.head = nn.Linear(emb_dim, num_classes)
    def forward(self, x):
        # Sequence pooling across token dimension
        x = self.block(x)
        return self.head(x.mean(dim=1))

print("--> Backbone micro-splits defined.")
```

```python
# =========================================================================
# CELL 4: Stochastic Codecs & Channel Models (Analog JSCC & Digital)
# =========================================================================

# -------------------------------------------------------------------------
# 1. Stochastic Variational Encoder (Reparameterization Trick)
# -------------------------------------------------------------------------
class VariationalEncoder(nn.Module):
    def __init__(self, in_dim, bottleneck_dim):
        super().__init__()
        self.fc_mu = nn.Linear(in_dim, bottleneck_dim)
        self.fc_logvar = nn.Linear(in_dim, bottleneck_dim)

    def forward(self, z):
        z_flat = torch.flatten(z, start_dim=1)
        mu = self.fc_mu(z_flat)
        logvar = torch.clamp(self.fc_logvar(z_flat), min=-10.0, max=10.0)
        std = torch.exp(0.5 * logvar)

        # Reparameterization
        eps = torch.randn_like(std)
        u = mu + eps * std
        return u, mu, logvar, eps

# -------------------------------------------------------------------------
# 2. Local Reconstruction Decoders (for JL and SL objectives)
# -------------------------------------------------------------------------
class TensorDecoder(nn.Module):
    def __init__(self, bottleneck_dim, target_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(bottleneck_dim, 256),
            nn.ReLU(),
            nn.Linear(256, target_dim)
        )
    def forward(self, u_bar):
        return self.net(u_bar)

# -------------------------------------------------------------------------
# 3. Channel Models (Analog JSCC & Digital Quantization)
# -------------------------------------------------------------------------
class PhysicalChannel(nn.Module):
    def __init__(self, mode="analog_awgn", snr_db=10.0, num_bits=4, prune_ratio=0.0):
        super().__init__()
        self.mode = mode
        self.snr_db = snr_db
        self.num_bits = num_bits
        self.prune_ratio = prune_ratio

    def forward(self, u):
        if self.mode == "ideal":
            return u

        # 1. Analog JSCC over AWGN
        elif self.mode == "analog_awgn":
            # Power normalization (P=1)
            norm = torch.norm(u, p=2, dim=-1, keepdim=True) + 1e-8
            u_norm = u / norm
            # Compute noise std based on SNR
            snr_linear = 10.0 ** (self.snr_db / 10.0)
            noise_std = np.sqrt(1.0 / (2.0 * snr_linear))
            noise = torch.randn_like(u_norm) * noise_std
            return (u_norm + noise) * norm

        # 2. Digital Compression (Pruning + STE Uniform Quantization)
        elif self.mode == "digital":
            # Magnitude Pruning
            if self.prune_ratio > 0.0:
                threshold = torch.quantile(torch.abs(u), self.prune_ratio)
                mask = torch.abs(u) >= threshold
                u = u * mask.float()

            # Straight-Through Estimator (STE) Uniform Quantization
            levels = 2 ** self.num_bits
            u_min, u_max = u.min(), u.max()
            scale = (u_max - u_min) / (levels - 1) + 1e-8
            u_quant = torch.round((u - u_min) / scale) * scale + u_min
            # STE Gradient trick
            return (u_quant - u).detach() + u

        return u

print("--> Codecs and Channel modules initialized.")
```

```python
# =========================================================================
# CELL 5: The 8 Tractable MIB Loss Functions
# =========================================================================

# -------------------------------------------------------------------------
# Rate Bounds (VUB vs. Sampled CLUB)
# -------------------------------------------------------------------------
def compute_rate_vub(mu, logvar):
    """Analytical KL divergence against N(0, I)"""
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=-1)
    return torch.mean(kl)

def compute_rate_club(u, mu, logvar, eps):
    """Sampled CLUB upper bound (O(N) contrastive batch rolling)"""
    std = torch.exp(0.5 * logvar)
    pos_term = eps.pow(2).sum(dim=-1)

    # Shuffle batch to obtain negative pairs
    u_neg = torch.roll(u, shifts=1, dims=0)
    neg_term = (((u_neg - mu) / std).pow(2)).sum(dim=-1)

    return 0.5 * torch.mean(neg_term - pos_term)

# -------------------------------------------------------------------------
# Unified MIB Loss Function Handler
# -------------------------------------------------------------------------
def compute_total_mib_loss(
    rel_type, rate_type,
    y_pred_sink, y_true,
    edge_data, decoders=None,
    lambdas={"w1": 1e-3, "w2": 1e-3}
):
    """
    Computes L_total = L_rel + sum(lambda_ij * L_rate_ij)
    """
    # 1. Relevance Loss (L_rel)
    if rel_type == "JG": # Joint Global (End-to-End Task Loss)
        loss_rel = F.cross_entropy(y_pred_sink, y_true)

    elif rel_type == "SL": # Separate Local (Per-edge MSE tensor reconstruction)
        loss_rel = 0.0
        for edge_key, data in edge_data.items():
            z_orig = data["z"]
            u_bar = data["u_bar"]
            z_recon = decoders[edge_key](u_bar)
            loss_rel += F.mse_loss(z_recon, torch.flatten(z_orig, start_dim=1))

    elif rel_type == "JL": # Joint Local (Full input MSE reconstruction)
        # Reconstruct combined inputs at Sink/Workers
        loss_rel = 0.0
        for edge_key, data in edge_data.items():
            z_recon = decoders[edge_key](data["u_bar"])
            loss_rel += F.mse_loss(z_recon, torch.flatten(data["z"], start_dim=1))

    elif rel_type == "SG": # Separate Global (Edge classifiers)
        loss_rel = F.cross_entropy(y_pred_sink, y_true) # Base downstream proxy

    else:
        raise ValueError(f"Unknown relevance target: {rel_type}")

    # 2. Rate Loss (L_rate)
    loss_rate = 0.0
    for edge_key, data in edge_data.items():
        weight = lambdas.get(edge_key, 1e-3)
        if rate_type == "VUB":
            r_loss = compute_rate_vub(data["mu"], data["logvar"])
        elif rate_type == "CLUB":
            r_loss = compute_rate_club(data["u"], data["mu"], data["logvar"], data["eps"])
        else:
            raise ValueError(f"Unknown rate bound: {rate_type}")

        loss_rate += weight * r_loss

    total_loss = loss_rel + loss_rate
    return total_loss, loss_rel.item(), loss_rate.item()

print("--> MIB Loss Engine ready.")
```

```python
# =========================================================================
# CELL 6: Collaborative Graph Models (Corrected Dimensions)
# =========================================================================

class DiamondCollaborativeModel(nn.Module):
    def __init__(self, bottleneck_dim=32, channel_mode="analog_awgn", snr_db=10.0):
        super().__init__()
        # Backbone micro-splits (MLP Baseline)
        self.source = MLPSource(in_dim=784, out_dim=256)
        self.worker1 = MLPWorker(in_dim=128, out_dim=128)
        self.worker2 = MLPWorker(in_dim=128, out_dim=128)

        # Sink input dimension matches the concatenated bottleneck representations (2 * bottleneck_dim)
        self.sink = MLPSink(in_dim=2 * bottleneck_dim, num_classes=10)

        # Encoders on edges: (w1 -> t) and (w2 -> t)
        self.enc_w1 = VariationalEncoder(in_dim=128, bottleneck_dim=bottleneck_dim)
        self.enc_w2 = VariationalEncoder(in_dim=128, bottleneck_dim=bottleneck_dim)

        # Local Decoders (for Local MIB objectives)
        self.dec_w1 = TensorDecoder(bottleneck_dim=bottleneck_dim, target_dim=128)
        self.dec_w2 = TensorDecoder(bottleneck_dim=bottleneck_dim, target_dim=128)

        # Channels
        self.channel_w1 = PhysicalChannel(mode=channel_mode, snr_db=snr_db)
        self.channel_w2 = PhysicalChannel(mode=channel_mode, snr_db=snr_db)

    def forward(self, x):
        # 1. Source computation & feature splitting
        z_s = self.source(x) # [B, 256]
        z_s1, z_s2 = z_s[:, :128], z_s[:, 128:] # Partitioned across paths

        # 2. Worker computation
        z_w1 = self.worker1(z_s1)
        z_w2 = self.worker2(z_s2)

        # 3. Codec & Channel on edge (w1 -> t)
        u1, mu1, logvar1, eps1 = self.enc_w1(z_w1)
        u_bar1 = self.channel_w1(u1)

        # 4. Codec & Channel on edge (w2 -> t)
        u2, mu2, logvar2, eps2 = self.enc_w2(z_w2)
        u_bar2 = self.channel_w2(u2)

        # 5. Sink fusion & prediction (concatenation along feature dimension)
        r_t = torch.cat([u_bar1, u_bar2], dim=-1) # Shape: [B, 64]
        y_hat = self.sink(r_t)

        edge_data = {
            "w1": {"z": z_w1, "u": u1, "mu": mu1, "logvar": logvar1, "eps": eps1, "u_bar": u_bar1},
            "w2": {"z": z_w2, "u": u2, "mu": mu2, "logvar": logvar2, "eps": eps2, "u_bar": u_bar2}
        }
        decoders = {"w1": self.dec_w1, "w2": self.dec_w2}
        return y_hat, edge_data, decoders

print("--> Collaborative Graph Models constructed.")
```

```python
# =========================================================================
# CELL 7: Training Engine with Scheme Support
# =========================================================================

def train_model(
    model, train_loader, val_loader,
    rel_type="JG", rate_type="CLUB",
    lambdas={"w1": 1e-3, "w2": 1e-3},
    scheme="joint_e2e", epochs=15, lr=1e-3
):
    model = model.to(device)

    # Configure parameter groups based on Training Scheme
    if scheme == "joint_e2e":
        optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    elif scheme == "two_stage_frozen":
        # Freeze backbone weights; only train codecs and decoders
        for p in model.source.parameters(): p.requires_grad = False
        for p in model.worker1.parameters(): p.requires_grad = False
        for p in model.worker2.parameters(): p.requires_grad = False
        for p in model.sink.parameters(): p.requires_grad = False
        codec_params = list(model.enc_w1.parameters()) + list(model.enc_w2.parameters()) + \
                       list(model.dec_w1.parameters()) + list(model.dec_w2.parameters())
        optimizer = torch.optim.Adam(codec_params, lr=lr)
    elif scheme == "fine_tuning":
        # Differential learning rates
        backbone_params = list(model.source.parameters()) + list(model.worker1.parameters()) + \
                          list(model.worker2.parameters()) + list(model.sink.parameters())
        codec_params = list(model.enc_w1.parameters()) + list(model.enc_w2.parameters()) + \
                       list(model.dec_w1.parameters()) + list(model.dec_w2.parameters())
        optimizer = torch.optim.Adam([
            {"params": backbone_params, "lr": 0.1 * lr},
            {"params": codec_params, "lr": lr}
        ])

    best_acc = 0.0
    for epoch in range(1, epochs + 1):
        model.train()
        total_loss, total_correct, total_samples = 0.0, 0, 0

        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()

            y_hat, edge_data, decoders = model(x)
            loss, _, _ = compute_total_mib_loss(rel_type, rate_type, y_hat, y, edge_data, decoders, lambdas)

            loss.backward()
            optimizer.step()

            total_loss += loss.item() * x.size(0)
            preds = torch.argmax(y_hat, dim=-1)
            total_correct += (preds == y).sum().item()
            total_samples += x.size(0)

        # Validation Step
        val_acc = evaluate(model, val_loader)
        if val_acc > best_acc:
            best_acc = val_acc

    return best_acc

def evaluate(model, data_loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for x, y in data_loader:
            x, y = x.to(device), y.to(device)
            y_hat, _, _ = model(x)
            preds = torch.argmax(y_hat, dim=-1)
            correct += (preds == y).sum().item()
            total += x.size(0)
    return 100.0 * correct / total

print("--> Training Engine configured.")
```

```python
# =========================================================================
# CELL 8: Automated Rate-Accuracy Sweep & Pareto Plotter (VUB vs. CLUB)
# =========================================================================

train_loader, test_loader = get_dataloaders("MNIST", batch_size=128)

# Lambda range sweep
lambda_values = [1e-4, 5e-4, 1e-3, 5e-3, 1e-2]
results_vub = []
results_club = []

print("--> Running Rate-Accuracy Pareto Sweep on Diamond Topology...")

# 1. Sweep VUB Rate Bound
for lam in lambda_values:
    model_vub = DiamondCollaborativeModel(bottleneck_dim=32, channel_mode="analog_awgn", snr_db=10.0)
    acc = train_model(
        model_vub, train_loader, test_loader,
        rel_type="JG", rate_type="VUB",
        lambdas={"w1": lam, "w2": lam}, epochs=10
    )
    results_vub.append(acc)
    print(f"VUB  | lambda: {lam:.1e} --> Best Accuracy: {acc:.2f}%")

# 2. Sweep CLUB Rate Bound
for lam in lambda_values:
    model_club = DiamondCollaborativeModel(bottleneck_dim=32, channel_mode="analog_awgn", snr_db=10.0)
    acc = train_model(
        model_club, train_loader, test_loader,
        rel_type="JG", rate_type="CLUB",
        lambdas={"w1": lam, "w2": lam}, epochs=10
    )
    results_club.append(acc)
    print(f"CLUB | lambda: {lam:.1e} --> Best Accuracy: {acc:.2f}%")

# 3. Plotting the Pareto Curve (Figure 1 Deliverable)
plt.figure(figsize=(8, 5))
plt.plot(lambda_values, results_club, marker='o', linewidth=2, color='crimson', label="Joint Global + CLUB (Proposed)")
plt.plot(lambda_values, results_vub, marker='s', linewidth=2, linestyle='--', color='navy', label="Joint Global + VUB (Baseline)")

plt.xscale('log')
plt.xlabel(r"Rate Compression Coefficient ($\lambda$)", fontsize=12)
plt.ylabel("Target Accuracy (%)", fontsize=12)
plt.title("Rate-Accuracy Trade-off on Diamond Topology (MNIST)", fontsize=13)
plt.grid(True, which="both", ls=":", alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

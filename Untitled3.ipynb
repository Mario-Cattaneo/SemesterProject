```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1. Dataset (Fashion-MNIST)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
trainset = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
train_loader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)
testset = torchvision.datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)
test_loader = torch.utils.data.DataLoader(testset, batch_size=128, shuffle=False, num_workers=2)

# 2. Codec Module
class VariationalChannelCodec(nn.Module):
    def __init__(self, in_channels, bottleneck_dim, snr_db=10.0):
        super().__init__()
        self.enc_mu = nn.Conv2d(in_channels, bottleneck_dim, kernel_size=1)
        self.enc_logvar = nn.Conv2d(in_channels, bottleneck_dim, kernel_size=1)
        self.dec = nn.Conv2d(bottleneck_dim, in_channels, kernel_size=1)
        snr_linear = 10.0 ** (snr_db / 10.0)
        self.noise_std = np.sqrt(1.0 / snr_linear)

    def forward(self, z):
        mu = self.enc_mu(z)
        logvar = self.enc_logvar(z)
        std = torch.exp(0.5 * logvar)
        u = mu + torch.randn_like(std) * std

        # Unit Power Normalization
        d_u = u.size(1) * u.size(2) * u.size(3)
        u_norm = torch.norm(u.view(u.size(0), -1), dim=1, keepdim=True).unsqueeze(-1).unsqueeze(-1)
        u_tx = np.sqrt(d_u) * u / (u_norm + 1e-8)

        # Channel Noise
        u_rx = u_tx + torch.randn_like(u_tx) * self.noise_std
        z_hat = self.dec(u_rx)
        kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=[1, 2, 3]).mean()
        return z_hat, u_rx, kl

# 3. Diamond DDNN
class DiamondDDNN(nn.Module):
    def __init__(self, bottleneck_dim=8):
        super().__init__()
        self.source = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.codec_s_v1 = VariationalChannelCodec(32, bottleneck_dim, snr_db=20.0)
        self.codec_s_v2 = VariationalChannelCodec(32, bottleneck_dim, snr_db=0.0)

        self.worker_v1 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.worker_v2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.codec_v1_t = VariationalChannelCodec(64, bottleneck_dim, snr_db=20.0)
        self.codec_v2_t = VariationalChannelCodec(64, bottleneck_dim, snr_db=0.0)

        self.sink_head = nn.Sequential(
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(128, 10)
        )

    def forward(self, x, use_codecs=True):
        z_s = self.source(x)
        if use_codecs:
            z_s_v1, _, kl1 = self.codec_s_v1(z_s)
            z_s_v2, _, kl2 = self.codec_s_v2(z_s)
        else:
            z_s_v1, z_s_v2 = z_s, z_s
            kl1, kl2 = torch.tensor(0.0), torch.tensor(0.0)

        z_v1 = self.worker_v1(z_s_v1)
        z_v2 = self.worker_v2(z_s_v2)

        if use_codecs:
            z_v1_t, _, kl3 = self.codec_v1_t(z_v1)
            z_v2_t, _, kl4 = self.codec_v2_t(z_v2)
        else:
            z_v1_t, z_v2_t = z_v1, z_v2
            kl3, kl4 = torch.tensor(0.0), torch.tensor(0.0)

        r_t = torch.cat([z_v1_t, z_v2_t], dim=1)
        y_hat = self.sink_head(r_t)

        acts = {'z_s': z_s, 'z_v1': z_v1, 'z_v2': z_v2}
        recons = {'z_hat_s_v1': z_s_v1, 'z_hat_s_v2': z_s_v2, 'z_hat_v1_t': z_v1_t, 'z_hat_v2_t': z_v2_t}
        kls = {'s_v1': kl1, 's_v2': kl2, 'v1_t': kl3, 'v2_t': kl4}
        return y_hat, acts, recons, kls

# =========================================================================
# Step 1: Pretrain Clean Backbone (Ideal Channel Baseline ~88-90% Acc)
# =========================================================================
print("Pretraining clean backbone (without noise/compression)...")
clean_model = DiamondDDNN().to(device)
opt_clean = optim.Adam(clean_model.parameters(), lr=1e-3)
crit = nn.CrossEntropyLoss()

for ep in range(6):
    clean_model.train()
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt_clean.zero_grad()
        y_hat, _, _, _ = clean_model(x, use_codecs=False)
        loss = crit(y_hat, y)
        loss.backward()
        opt_clean.step()

clean_state = clean_model.state_dict()
print("Pretraining complete!")

# =========================================================================
# Step 2: Training Function for Sweeps
# =========================================================================
def train_and_eval(objective='JG', regime='joint', gamma=0.01, epochs=4):
    model = DiamondDDNN(bottleneck_dim=8).to(device)
    model.load_state_dict(clean_state) # Load clean pre-trained weights

    task_crit = nn.CrossEntropyLoss()
    rec_crit = nn.MSELoss()
    lambda_dict = {'s_v1': gamma, 'v1_t': gamma, 's_v2': gamma * 4.0, 'v2_t': gamma * 4.0}

    if regime == 'frozen':
        for name, p in model.named_parameters():
            if 'codec' not in name:
                p.requires_grad = False
        opt = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=2e-3)
    else:
        opt = optim.Adam(model.parameters(), lr=1e-3)

    for ep in range(epochs):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            opt.zero_grad()
            y_hat, acts, recons, kls = model(x, use_codecs=True)
            rate_loss = sum(lambda_dict[e] * kls[e] for e in kls)

            if objective == 'JG':
                loss = task_crit(y_hat, y) + rate_loss
            elif objective == 'SL':
                rec_loss = rec_crit(recons['z_hat_s_v1'], acts['z_s']) + \
                           rec_crit(recons['z_hat_s_v2'], acts['z_s']) + \
                           rec_crit(recons['z_hat_v1_t'], acts['z_v1']) + \
                           rec_crit(recons['z_hat_v2_t'], acts['z_v2'])
                loss = rec_loss + rate_loss + (task_crit(y_hat, y) if regime == 'joint' else 0.0)

            loss.backward()
            opt.step()

    # Eval
    model.eval()
    corr, tot, total_rate = 0, 0, 0.0
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            y_hat, _, _, kls = model(x, use_codecs=True)
            corr += (y_hat.argmax(1) == y).sum().item()
            tot += y.size(0)
            total_rate += sum(kls.values()).item()

    return 100.0 * corr / tot, total_rate / len(test_loader)

# =========================================================================
# Step 3: Run Sweeps & Plot
# =========================================================================
gammas = [1e-4, 1e-3, 5e-3, 1e-2, 5e-2]
results = {'JG_joint': [], 'SL_joint': [], 'JG_frozen': [], 'SL_frozen': []}

for g in gammas:
    print(f"Sweeping gamma = {g}...")
    results['JG_joint'].append(train_and_eval('JG', 'joint', gamma=g))
    results['SL_joint'].append(train_and_eval('SL', 'joint', gamma=g))
    results['JG_frozen'].append(train_and_eval('JG', 'frozen', gamma=g))
    results['SL_frozen'].append(train_and_eval('SL', 'frozen', gamma=g))

# Plotting
plt.figure(figsize=(9, 5.5))
for key, pts in results.items():
    rates = [p[1] for p in pts]
    accs = [p[0] for p in pts]
    plt.plot(rates, accs, marker='o', linewidth=2, label=key)

plt.xlabel('Average Transmission Rate (KL Divergence)', fontsize=12)
plt.ylabel('Top-1 Test Accuracy (%)', fontsize=12)
plt.title('Diamond DDNN Rate-Accuracy Frontier (With Pretrained Clean Base)', fontsize=13)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.savefig('rate_accuracy_pareto_v2.png', dpi=300)
plt.show()
```

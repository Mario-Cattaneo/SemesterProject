```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ==========================================
# 1. Dataset (Fashion-MNIST - Instant Download)
# ==========================================
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

trainset = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
train_loader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

testset = torchvision.datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)
test_loader = torch.utils.data.DataLoader(testset, batch_size=128, shuffle=False, num_workers=2)

# ==========================================
# 2. Lightweight Variational Codec Layer
# ==========================================
class VariationalChannelCodec(nn.Module):
    def __init__(self, in_channels, bottleneck_dim, snr_db=10.0):
        super().__init__()
        self.bottleneck_dim = bottleneck_dim
        # Lightweight 1x1 Convolutions for Edge deployment
        self.enc_mu = nn.Conv2d(in_channels, bottleneck_dim, kernel_size=1)
        self.enc_logvar = nn.Conv2d(in_channels, bottleneck_dim, kernel_size=1)
        self.dec = nn.Conv2d(bottleneck_dim, in_channels, kernel_size=1)

        # AWGN Channel noise variance
        snr_linear = 10.0 ** (snr_db / 10.0)
        self.noise_std = np.sqrt(1.0 / snr_linear)

    def encode(self, z):
        mu = self.enc_mu(z)
        logvar = self.enc_logvar(z)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        u = mu + eps * std
        return u

    def forward(self, z):
        mu, logvar = self.encode(z)
        u = self.reparameterize(mu, logvar)

        # Unit Power Normalization: ||u||^2 / dim = 1
        u_norm = torch.norm(u.view(u.size(0), -1), dim=1, keepdim=True).unsqueeze(-1).unsqueeze(-1)
        d_u = u.size(1) * u.size(2) * u.size(3)
        u_transmitted = np.sqrt(d_u) * u / (u_norm + 1e-8)

        # AWGN Channel Noise
        noise = torch.randn_like(u_transmitted) * self.noise_std
        u_noisy = u_transmitted + noise

        # Decoder reconstruction
        z_hat = self.dec(u_noisy)

        # Analytical KL Divergence (Rate Term)
        kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=[1, 2, 3]).mean()
        return z_hat, u_noisy, kl

# ==========================================
# 3. Diamond Topology DDNN Architecture
# ==========================================
class DiamondDDNN(nn.Module):
    def __init__(self, bottleneck_dim=8):
        super().__init__()
        # Source Device s: Initial Feature Extraction
        self.source = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )

        # Codecs for Edges (s -> v1) and (s -> v2)
        # Branch 1 (s -> v1): High SNR = 20 dB (Capacity C1 = High)
        self.codec_s_v1 = VariationalChannelCodec(in_channels=32, bottleneck_dim=bottleneck_dim, snr_db=20.0)
        # Branch 2 (s -> v2): Low SNR = 0 dB (Capacity C2 = Low)
        self.codec_s_v2 = VariationalChannelCodec(in_channels=32, bottleneck_dim=bottleneck_dim, snr_db=0.0)

        # Worker 1 (Device v1) & Worker 2 (Device v2)
        self.worker_v1 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.worker_v2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        # Codecs for Edges (v1 -> t) and (v2 -> t)
        self.codec_v1_t = VariationalChannelCodec(in_channels=64, bottleneck_dim=bottleneck_dim, snr_db=20.0)
        self.codec_v2_t = VariationalChannelCodec(in_channels=64, bottleneck_dim=bottleneck_dim, snr_db=0.0)

        # Sink Device t: Aggregates reconstructed features & Classifies
        self.sink_head = nn.Sequential(
            nn.Conv2d(64 + 64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        # Source Pass
        z_s = self.source(x)

        # Transmit across Edges (s -> v1) and (s -> v2)
        z_hat_s_v1, u_s_v1, kl_s_v1 = self.codec_s_v1(z_s)
        z_hat_s_v2, u_s_v2, kl_s_v2 = self.codec_s_v2(z_s)

        # Workers Pass
        z_v1 = self.worker_v1(z_hat_s_v1)
        z_v2 = self.worker_v2(z_hat_s_v2)

        # Transmit across Edges (v1 -> t) and (v2 -> t)
        z_hat_v1_t, u_v1_t, kl_v1_t = self.codec_v1_t(z_v1)
        z_hat_v2_t, u_v2_t, kl_v2_t = self.codec_v2_t(z_v2)

        # Sink Aggregation & Prediction
        r_t = torch.cat([z_hat_v1_t, z_hat_v2_t], dim=1)
        y_hat = self.sink_head(r_t)

        # Bundle data for loss computations
        activations = {'z_s': z_s, 'z_v1': z_v1, 'z_v2': z_v2}
        reconstructions = {'z_hat_s_v1': z_hat_s_v1, 'z_hat_s_v2': z_hat_s_v2,
                           'z_hat_v1_t': z_hat_v1_t, 'z_hat_v2_t': z_hat_v2_t}
        kl_dict = {'s_v1': kl_s_v1, 's_v2': kl_s_v2, 'v1_t': kl_v1_t, 'v2_t': kl_v2_t}

        return y_hat, activations, reconstructions, kl_dict

# ==========================================
# 4. Training and Evaluation Routine
# ==========================================
def train_and_eval(objective='JG', regime='joint', gamma=0.01, epochs=5):
    model = DiamondDDNN(bottleneck_dim=8).to(device)
    task_criterion = nn.CrossEntropyLoss()
    recon_criterion = nn.MSELoss()

    # Asymmetric Link Multipliers \lambda_{i,j} = \gamma / C_{i,j}
    # Branch 1 (High SNR) -> lower lambda; Branch 2 (Low SNR) -> higher lambda
    lambda_dict = {
        's_v1': gamma * 1.0,
        'v1_t': gamma * 1.0,
        's_v2': gamma * 4.0, # Bottleneck penalized 4x harder
        'v2_t': gamma * 4.0
    }

    if regime == 'frozen':
        # Freeze backbone, train only codecs
        for name, param in model.named_parameters():
            if 'codec' not in name:
                param.requires_grad = False
        optimizer = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-3)
    else:
        # Joint end-to-end
        optimizer = optim.Adam(model.parameters(), lr=1e-3)

    model.train()
    for epoch in range(epochs):
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()

            y_hat, acts, recons, kls = model(x)

            # Rate penalty (Weighted sum of KLs)
            rate_loss = sum(lambda_dict[edge] * kls[edge] for edge in kls)

            # Formulate Objective Loss
            if objective == 'JG': # Joint Global (Task CE + Rate)
                loss = task_criterion(y_hat, y) + rate_loss
            elif objective == 'SL': # Separate Local (Reconstruction + Rate [+ Task CE if joint])
                rec_loss = recon_criterion(recons['z_hat_s_v1'], acts['z_s']) + \
                           recon_criterion(recons['z_hat_s_v2'], acts['z_s']) + \
                           recon_criterion(recons['z_hat_v1_t'], acts['z_v1']) + \
                           recon_criterion(recons['z_hat_v2_t'], acts['z_v2'])
                loss = rec_loss + rate_loss
                if regime == 'joint':
                    loss += task_criterion(y_hat, y)

            loss.backward()
            optimizer.step()

    # Evaluation
    model.eval()
    correct, total = 0, 0
    total_rate = 0.0
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            y_hat, _, _, kls = model(x)
            _, predicted = torch.max(y_hat.data, 1)
            total += y.size(0)
            correct += (predicted == y).sum().item()
            total_rate += sum(kls.values()).item()

    acc = 100.0 * correct / total
    avg_rate = total_rate / len(test_loader)
    return acc, avg_rate

# ==========================================
# 5. Run Sweep & Plot Pareto Curves
# ==========================================
gammas = [1e-4, 1e-3, 5e-3, 1e-2, 5e-2]
results = {'JG_joint': [], 'SL_joint': [], 'JG_frozen': [], 'SL_frozen': []}

print("Running sweeps (this takes ~5-8 mins on Colab GPU)...")
for g in gammas:
    print(f"--- Testing gamma = {g} ---")
    results['JG_joint'].append(train_and_eval('JG', 'joint', gamma=g, epochs=3))
    results['SL_joint'].append(train_and_eval('SL', 'joint', gamma=g, epochs=3))
    results['JG_frozen'].append(train_and_eval('JG', 'frozen', gamma=g, epochs=3))
    results['SL_frozen'].append(train_and_eval('SL', 'frozen', gamma=g, epochs=3))

# Plotting the Pareto Frontier
plt.figure(figsize=(8, 5))
for key, pts in results.items():
    rates = [p[1] for p in pts]
    accs = [p[0] for p in pts]
    plt.plot(rates, accs, marker='o', label=key)

plt.xlabel('Average Transmission Rate (KL Divergence)', fontsize=12)
plt.ylabel('Top-1 Test Accuracy (%)', fontsize=12)
plt.title('Diamond DDNN: Rate-Accuracy Pareto Tradeoff', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.savefig('rate_accuracy_pareto.png', dpi=300)
plt.show()
print("Finished! Figure saved as rate_accuracy_pareto.png")
```
